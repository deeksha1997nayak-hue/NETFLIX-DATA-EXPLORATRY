## 2. Create a sample dataset
Each dictionary represents one title. Movie duration is in minutes; TV show duration is number of seasons.

```python
titles = [
    {"title": "Skybound", "type": "Movie", "year": 2021, "genre": "Action", "duration": 112},
    {"title": "Little Kitchen", "type": "TV Show", "year": 2020, "genre": "Comedy", "duration": 2},
    {"title": "Ocean Secrets", "type": "Movie", "year": 2019, "genre": "Documentary", "duration": 95},
    {"title": "City Detectives", "type": "TV Show", "year": 2022, "genre": "Drama", "duration": 3},
    {"title": "The Last Signal", "type": "Movie", "year": 2023, "genre": "Sci-Fi", "duration": 124},
    {"title": "Laugh Track", "type": "TV Show", "year": 2018, "genre": "Comedy", "duration": 4},
    {"title": "Mountain Rescue", "type": "Movie", "year": 2017, "genre": "Action", "duration": 103},
    {"title": "Hidden Garden", "type": "Movie", "year": 2022, "genre": "Drama", "duration": 118},
    {"title": "Planet Earthlings", "type": "TV Show", "year": 2021, "genre": "Documentary", "duration": 1},
    {"title": "Puzzle House", "type": "Movie", "year": 2016, "genre": "Thriller", "duration": 107},
    {"title": "Weekend Stories", "type": "TV Show", "year": 2023, "genre": "Drama", "duration": 2},
    {"title": "Star Harbor", "type": "Movie", "year": 2020, "genre": "Sci-Fi", "duration": 130}
]
print("Total sample titles:", len(titles))
for item in titles[:3]:
    print(item)
```

## 3. Strings and methods

```python
title = titles[0]["title"]
print("Original:", title)
print("Uppercase:", title.upper())
print("Lowercase:", title.lower())
print("Length:", len(title))
```

## 4. Lists, tuples, sets, and dictionaries

```python
title_names = [item["title"] for item in titles]  # list
first_details = (titles[0]["title"], titles[0]["year"], titles[0]["genre"])  # tuple
unique_genres = {item["genre"] for item in titles}  # set
first_record = titles[0]  # dictionary

print("Title list:", title_names)
print("First title tuple:", first_details)
print("Unique genres:", unique_genres)
print("First title dictionary:", first_record)
```

## 5. Conditionals: classify titles by release year

```python
def release_category(year):
    if year >= 2022:
        return "Recent (2022 or later)"
    elif year >= 2019:
        return "2019–2021"
    else:
        return "Before 2019"

for item in titles:
    print(item["title"], "->", release_category(item["year"]))
```

## 6. Loops: count movies and TV shows

```python
movie_count = 0
tv_count = 0
for item in titles:
    if item["type"] == "Movie":
        movie_count += 1
    elif item["type"] == "TV Show":
        tv_count += 1

print("Movies:", movie_count)
print("TV shows:", tv_count)
print("Total:", movie_count + tv_count)
```

## 7. List comprehension: recent titles

```python
recent_titles = [item["title"] for item in titles if item["year"] >= 2022]
print("Titles from 2022 or later:")
for name in recent_titles:
    print("-", name)
```

## 8. Functions: calculate simple statistics

```python
def count_by_type(dataset, content_type):
    return sum(1 for item in dataset if item["type"] == content_type)

def average_movie_duration(dataset):
    durations = [item["duration"] for item in dataset if item["type"] == "Movie"]
    return sum(durations) / len(durations) if durations else 0

print("Movies:", count_by_type(titles, "Movie"))
print("TV shows:", count_by_type(titles, "TV Show"))
print("Average movie duration:", round(average_movie_duration(titles), 1), "minutes")
```

## 9. Dictionary analysis: count titles by genre

```python
genre_counts = {}
for item in titles:
    genre = item["genre"]
    genre_counts[genre] = genre_counts.get(genre, 0) + 1

for genre, count in genre_counts.items():
    print(f"{genre}: {count}")
```

## 10. Set intersection: genres shared by movies and TV shows

```python
movie_genres = [item["genre"] for item in titles if item["type"] == "Movie"]
show_genres = [item["genre"] for item in titles if item["type"] == "TV Show"]
common_genres = set(movie_genres) & set(show_genres)
print("Genres in both groups:", sorted(common_genres))
```

## 11. While loop

```python
step = 1
while step <= 3:
    print("Practice step", step)
    step += 1
print("Loop finished.")
```

## 12. Basic error handling

```python
def check_year(value):
    try:
        year = int(value)
    except (ValueError, TypeError):
        return "Invalid year: enter a whole number."
    if year < 1900 or year > 2100:
        return "Year is outside the expected range."
    return f"Valid year: {year}"

print(check_year(2023))
print(check_year("hello"))
print(check_year(2500))
```

## 13. Simple text-based charts

```python
print("Content type counts")
for label, count in [("Movies", movie_count), ("TV Shows", tv_count)]:
    print(f"{label:10} | {'#' * count} ({count})")

print("\nGenre counts")
for genre, count in sorted(genre_counts.items(), key=lambda pair: pair[1], reverse=True):
    print(f"{genre:14} | {'#' * count} ({count})")
```

## 14. Automatically summarize sample results

```python
most_common_genre = max(genre_counts, key=genre_counts.get)
movie_share = movie_count / len(titles) * 100
show_share = tv_count / len(titles) * 100

print(f"Total sample titles: {len(titles)}")
print(f"Movies: {movie_count} ({movie_share:.1f}%)")
print(f"TV shows: {tv_count} ({show_share:.1f}%)")
print(f"Most common sample genre: {most_common_genre} ({genre_counts[most_common_genre]} titles)")
print(f"Average movie duration: {average_movie_duration(titles):.1f} minutes")
print("These results describe only this small practice dataset.")
```
